```python


def replace_guards(var):
    """
	This in-place replaces the points in the guard cells with the points on the boundary
    
    """
    # Strip the edge guard cells
    var = var[1:-1]

    var[0] = 0.5*(var[0] + var[1])
    var[-1] = 0.5*(var[-1] + var[-2])
    return var
```

```python
import scipy.integrate

def divq_integrate(dataframe, snb_int = False):
    """
    Calculate the total heat flux from the divergence of the Spitzer-Harm fluxes.
    If snb == True outputs integral of divq_snb, otherwise outputs integral of divq_sh.
    """

    x = np.ravel(dataframe['y'].values)[1:-1]
    Te = np.ravel(dataframe['Te'].values)


    div_q_snb = replace_guards(np.ravel(dataframe['Div_Q_SNB'].values))
    div_q_sh = replace_guards(np.ravel(dataframe['Div_Q_SH'].values))

    print('x =' , len(x))
    print('div_q_snb = ' , len(div_q_snb))

    q_snb = scipy.integrate.cumulative_trapezoid(div_q_snb, x, initial=0)
    q_sh = scipy.integrate.cumulative_trapezoid(div_q_sh, x, initial=0)

    if snb_int == False:
        return q_sh
    else:
        return q_snb
```

```python
def spitzer_q(dataframe):
    
    x = dataframe['y']
    Te = dataframe['Te']
    Ti = dataframe['Td+']
    Ne = dataframe['Ne']
    Ni = dataframe['Nd+']
    kappa_e = dataframe['kappa_par_e']
    kappa_i = dataframe['kappa_par_d+']

    # print(kappa_e)

    grad_T = np.gradient(Te, x)
    q = -kappa_e * grad_T

    return q
```

```python
import numpy as np

def neon_adas_curve(Te):
    logT = np.log(Te)
    log_out = np.zeros_like(Te)

    # Coefficients for the polynomial part
    coefficients = np.array([
        -8.21475117e+01, 1.28929854e+01, -4.74266289e+01,
         7.45222324e+01, -5.75710722e+01, 2.57375965e+01,
        -7.12758563e+00, 1.24287546e+00, -1.32943407e-01,
         7.97368445e-03, -2.05487897e-04
    ])

    # Valid temperature range for the polynomial
    valid_mask = (Te >= 2) & (Te <= 1000)
    below_mask = Te < 2
    above_mask = Te > 1000

    # Apply the polynomial coefficients
    for i, coeff in enumerate(coefficients):
        log_out[valid_mask] += coeff * logT[valid_mask] ** i
    
    # Apply the exponential to the valid range
    log_out[valid_mask] = np.exp(log_out[valid_mask])
    
    # Assign the constant values outside the valid temperature range
    log_out[below_mask] = 6.35304113e-36
    log_out[above_mask] = 1.17894628e-32

    return log_out

# Example usage with an array of temperatures:
Te_array = np.array([1.0, 5.0, 10.0, 100.0, 500.0, 1500.0])
output_values = neon_adas_curve(Te_array)
# print(output_values)


# Use this updated Neon_adas class to create the color map and overlay it onto the plot

Te = np.linspace(0, 500, 1000)  # Example temperature value
result = neon_adas_curve(Te)

fig,axis = plt.subplots(1, 1, figsize=(10, 6), dpi = 500)
# axis2 = axis.twinx()


max_index = np.argmax(result)
print(max_index)

print(Te[max_index])

axis.plot(Te, result, label='Neon ADAS Curve', linewidth=linewidth, color = 'blue')
axis.axvspan(100, 200, color='darkgreen', alpha=0.3)  # Red color with transparency
axis.axvspan(12.5, 100, color='limegreen', alpha=0.3)  # Red color with transparency
axis.axvspan(200, 500, color='darkred', alpha=0.3)  # Red color with transparency
axis.axvspan(0, 12.5, color='red', alpha=0.3)  # Red color with transparency

# for i in alpha_vals:

#     if i == -1:
#         linestyle = '--'
#         color = 'red'
#     elif i == 0.2:
#         linestyle = '-'
#         color = 'blue'
#     elif i == 0.06:
#         linestyle = '-'
#         color = 'black'
#     elif i == 'snb':
#         linestyle = ':'
#         color = 'magenta'
#     axis2.plot(replace_guards(cs[f'alpha_{i}_neon_0.0'].ds.isel(t=-1)['Te'].values), replace_guards(cs[f'alpha_{i}_neon_0.0'].ds.isel(t=-1)['Ne'].values), label=f'alpha = {i}', linewidth=linewidth)
# axis2.set_ylabel(r'Electron density (m$^{-3}$)')
# axis2.legend(loc='upper right')
    
axis.set_xlabel('Te (eV)')
axis.set_ylabel(r'Aggregate Neon Radiation (Wm$^{3}$)')
axis.set_xbound(0, 500)
# axis.set_yscale('log')
# axis.set_xbound(0,200)
```

```python
import matplotlib.pyplot as plt
import numpy as np

linewidth = 2

def plot_neon_profiles(params, neon_value, markersize=6, x_scale = 'linear'):
    """
    Plots neon profiles for different parameters.

    Parameters:
      params (list): A list of parameter names (e.g., ['Te', 'q_cond', 'Ne', 'Rneon']).
      neon_value (str): The neon concentration value as a string (e.g., '0.0').
      markersize (int): The marker size for plotting (default is 6).

    This function assumes that global variables and functions like `cs`, `alpha_vals`,
    `replace_guards`, `divq_integrate`, and `spitzer_q` are defined elsewhere.
    """
    # Determine subplot grid dimensions
    ncols = max(round(len(params) / 2), 1)
    nrows = ncols
    print("Subplot grid dimensions:", ncols, nrows)

    fig, ax = plt.subplots(nrows, ncols, figsize=(10, 6), dpi=300)

    # Handle the case of a single subplot vs. multiple subplots
    if isinstance(ax, np.ndarray):
        ax = ax.flatten()  # Flatten the 2D array of axes to 1D
    else:
        ax = np.array([ax])  # Ensure ax is a 1D array for consistency

    # Get the 'y' coordinate array from one of the datasets.
    # Adjust the slicing as needed.
    y = cs[f'alpha_-1_neon_{neon_value}'].ds['y'].values[1:-1]
    y = cs['alpha_-1_neon_0.0'].ds['y'].values[1:-1]
    y_flipped = (y.max() - y) 
    # Loop over all alpha values
    for i in alpha_vals:
        target_temp_list = []
        neon_radiation_list = []
        
        # Define linestyle, color, and label based on the alpha value
        if i == -1:
            linestyle = '--'
            color = 'red'
            label = 'SH'
        elif i == 0.2:
            linestyle = '-'
            color = 'blue'
            label = i
        elif i == 0.06:
            linestyle = '-'
            color = 'black'
            label = i
        elif i == 'snb':
            linestyle = ':'
            color = 'magenta'
            label = i
        else:
            linestyle = '-'
            color = 'green'
            label = i

        # Loop over the parameters provided in the input list
        for j, value in enumerate(params):
            ds_key = f'alpha_{i}_neon_{neon_value}'
            if value == 'q_cond':
                if i == 'snb':
                    # Use the snb integral if alpha is 'snb'
                    profile = divq_integrate(cs[ds_key].ds.isel(t=-1), snb_int=True)
                else:
                    # Compute q_cond using spitzer_q (or another method)
                    profile = replace_guards(spitzer_q(cs[ds_key].ds.isel(t=-1)))
                # Plot the q_cond profile (scaling the units appropriately)
                ax[j].plot(y_flipped[3:-3], profile[3:-3] * 1e-6, linestyle=linestyle, 
                           color=color, label=f'alpha={label}', linewidth=linewidth)
                ax[j].set_ylabel(r'$q_{\parallel, e}$ (MW/m$^2$)')
                ax[j].set_xlabel('$S_{\parallel}$ (m)')

            elif value == 'Rneon':
                # Calculate neon radiation profile
                profile = -1 * replace_guards(cs[ds_key].ds.isel(t=-1)[value].values) * \
                          replace_guards(cs[ds_key].ds.isel(t=-1)['dv'].values)
                ax[j].plot(y_flipped[1:-1], profile[1:-1]*1e-6, linestyle=linestyle, color=color, 
                           label=f'alpha={label}', linewidth=linewidth)
                ax[j].set_xlabel('$S_{\parallel}$ (m)')
                units = cs[f'alpha_{i}_neon_{neon_value}'].ds.isel(t=-1)[value].units
                ax[j].set_ylabel(f'Neon radiation (MW)')

            elif value == 'Ne':
                # Plot the electron density profile
                profile = replace_guards(cs[ds_key].ds.isel(t=-1)[value].values)
                ax[j].plot(y_flipped, profile, linestyle=linestyle, color=color, label=f'alpha={label}', linewidth=linewidth)
                if np.max(profile) > 1e3:
                    ax[j].set_yscale('log')
                ax[j].set_xlabel('$S_{\parallel}$ (m)')
                units = cs[f'alpha_{i}_neon_{neon_value}'].ds.isel(t=-1)[value].units
                ax[j].set_ybound(1e19, 1e22)
                ax[j].set_ylabel(f'{value} ({units})')

            else:
                # For other parameters like 'Te' or 'Ne'
                profile = replace_guards(cs[ds_key].ds.isel(t=-1)[value].values)
                ax[j].plot(y_flipped, profile, linestyle=linestyle, color=color, label=f'alpha={label}', linewidth=linewidth)
                if np.max(profile) > 1e3:
                    ax[j].set_yscale('symlog')
                ax[j].set_xlabel('$S_{\parallel}$ (m)')
                units = cs[f'alpha_{i}_neon_{neon_value}'].ds.isel(t=-1)[value].units
                ax[j].set_ylabel(f'{value} ({units})')

            # Append values for potential further processing (if needed)
            # For example, you might collect target temperatures or neon radiation values here.

        # End for j (params loop)

    # Customize layout for all subplots
    plt.tight_layout()

    # Set x-scale for each axis
    for axes in ax:
        axes.set_xscale(x_scale)
        axes.set_xlabel('Parallel distance to target (m)')

    # Create a single legend for the entire figure using handles/labels from the first axis.
    handles, labels = ax[0].get_legend_handles_labels()
    fig.legend(handles, labels, loc='upper center', ncol=len(labels), bbox_to_anchor=(0.5, 1.07))
    fig.subplots_adjust(top=0.85)  # Adjust to make space for the legend

    # Set a title for the entire figure
    fig.suptitle(f'Neon concentration = {neon_value}', fontsize=16)

    plt.tight_layout(rect=[0, 0, 1, 0.98])  # Adjust layout to keep space for the legend

    plt.show()

# Example call for the function:
params = ['Te', 'q_cond', 'Ne', 'Rneon']
# params = ['Rneon']
neon_value = '0.032'
plot_neon_profiles(params, neon_value, x_scale='linear')
```
